## Connect to Database  

```python
from sqlalchemy import create_engine, text
from urllib.parse import quote_plus

password = quote_plus("")

# ── Step 1: amazon_db create karo ─────────────────────

engine_default = create_engine(
    f"postgresql://postgres:{password}@"
)

with engine_default.connect().execution_options(isolation_level="AUTOCOMMIT") as conn:
    try:
        conn.execute(text("CREATE DATABASE amazon_db"))
        print("✅ Database 'amazon_db' created!")
    except Exception:
        print("ℹ️ Database already exists, moving on...")

engine_default.dispose()

# ── Step 2: amazon_db se connect karo ─────────────────

engine = create_engine(
    f"postgresql://postgres:{password}@localhost:5432/amazon_db"
)

with engine.connect() as conn:
    print("✅ Connected to amazon_db!")

# ── Step 3: Purani tables drop karo ───────────────────

with engine.begin() as conn:
    conn.execute(text("DROP TABLE IF EXISTS revenue CASCADE"))
    conn.execute(text("DROP TABLE IF EXISTS orders CASCADE"))
    conn.execute(text("DROP TABLE IF EXISTS customers CASCADE"))

print("✅ Old tables dropped!")

# ── Step 4: Cleaned data save karo ────────────────────

customer_df.to_sql("customers", engine, if_exists="replace", index=False)
print("✅ customers saved")

orders_df.to_sql("orders", engine, if_exists="replace", index=False)
print("✅ orders saved")

revenue_df.to_sql("revenue", engine, if_exists="replace", index=False)
print("✅ revenue saved")

# ── Step 5: Verify karo ───────────────────────────────

with engine.connect() as conn:
    for table in ["customers", "orders", "revenue"]:
        count = conn.execute(text(f"SELECT COUNT(*) FROM {table}")).scalar()
        print(f"✅ {table}: {count} rows")

engine.dispose()

print("\n🎉 Done! Amazon data PostgreSQL mein save ho gaya.")
```
